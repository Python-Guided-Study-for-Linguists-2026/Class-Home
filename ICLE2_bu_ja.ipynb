### Indicators of writing quality. TTR: Type-Token Ratio
- Type-token ratio (TTR) is a quick measurement for vocabulary diversity.
- Per-group measurement and also per-essay measurement.
- But there's a big caveat. What? 

```python
# per-group measurements
print('BU:', len(bucor_fd)/len(bucor.words()))
print('JA:', len(jacor_fd)/len(jacor.words()))
```

```python
def get_ttr(wlist):
    wlist_lower = [w.lower() for w in wlist]
    return round(len(set(wlist_lower))/len(wlist_lower), 4)

get_ttr('Rose is a rose is a rose is a rose'.split())
```

```python
bucor_ttr = [get_ttr(bucor.words(f)) for f in bucor.fileids()]
jacor_ttr = [get_ttr(jacor.words(f)) for f in jacor.fileids()]
print('BU:', sorted(bucor_ttr))
print('JA:', sorted(jacor_ttr))
```

- OK, those are the numbers... but _how to interpret them_? 
- Also, the big CAVEAT? What do we do about it?

### Indicators of writing quality. Average word length

```python
def get_avg_wlen(wlist):
    wlen = [len(w) for w in wlist]
    return round(sum(wlen)/len(wlen), 2)

get_avg_wlen('Rose is a rose is a rose is a rose'.split())
```

```python
# per-group measurements
print('BU:', get_avg_wlen(bucor.words()))
print('JA:', get_avg_wlen(jacor.words()))
```

```python
bucor_wlen = [get_avg_wlen(bucor.words(f)) for f in bucor.fileids()]
jacor_wlen = [get_avg_wlen(jacor.words(f)) for f in jacor.fileids()]
print('BU:', sorted(bucor_wlen))
print('JA:', sorted(jacor_wlen))
```

#### TTR, average word length
- For TTR, let's create an improved function that lets us use truncated portion 

```python
def get_ttr_trunc(wlist, n): 
    # truncated TTR, computes TTR on first n word tokens 
    wlist_lower = [w.lower() for w in wlist][:n]   # first n lowercased tokens
    return round(len(set(wlist_lower))/len(wlist_lower), 4)

get_ttr_trunc('Rose is a rose is a rose is a rose'.split(), 100)
```

```python
bucor_ttr = [get_ttr_trunc(bucor.words(f), 100) for f in bucor.fileids()]
jacor_ttr = [get_ttr_trunc(jacor.words(f), 100) for f in jacor.fileids()]
print('BU:', sorted(bucor_ttr))
print('    avg:', statistics.mean(bucor_ttr))
print('JA:', sorted(jacor_ttr))
print('    avg:', statistics.mean(jacor_ttr))
```

```python
inaug_recent_ttr = [get_ttr_trunc(inaug_recent.words(f), 100) for f in inaug_recent.fileids()]
print('US presidents:', sorted(inaug_recent_ttr)) 
print('    avg:', statistics.mean(inaug_recent_ttr))
```

```python
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
# a text for testing 
bu_essay = bucor.raw(bucor.fileids()[0])
print(bu_essay)
```

```python
# let's make it shorter
bu_essay = bu_essay[:bu_essay.index('Children')]
bu_essay
```

```python
doc1 = nlp(bu_essay)
```

```python
for token in doc1:
    print(token.text, token.lemma_, token.pos_, token.tag_, token.dep_,
          token.is_stop, sep="\t")
```

```python
# What can you do with the document object?
dir(doc1)
```

```python
# noun phrases
for chunk in doc1.noun_chunks:
    print(chunk.text, chunk.root.text, chunk.root.dep_, chunk.root.head.text, sep="\t\t")
```

```python
# tokenized sentences
for sent in doc1.sents:
    print(sent)
```

```python
# Let's take a look at syntactic trees. 
# spaCy uses "dependency trees", different from our usual "constituent trees"
from spacy import displacy

for sent in doc1.sents:
    displacy.render(sent, style='dep')
    # displacy.render(sent, style='dep', options={'compact':True})  # for compact style
```
