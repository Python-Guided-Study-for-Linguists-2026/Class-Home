### More corpus stats
- sentence counts and average sentence lengths (= # of words in sentence)
- measurement per group and per individual essays

```python
# corpus sentence counts, average sentence count per essay
print('BU:', len(bucor.sents()), len(bucor.sents())/30)
print('JA:', len(jacor.sents()), len(jacor.sents())/30)
```

```python
# average sentence lengths in # of words
print('BU:', len(bucor.words())/len(bucor.sents()))
print('JA:', len(jacor.words())/len(jacor.sents()))
```

```python
# sentence count, but per-essay. 
bucor_sentcount = [len(bucor.sents(f)) for f in bucor.fileids()]
jacor_sentcount = [len(jacor.sents(f)) for f in jacor.fileids()]
print('BU:', sorted(bucor_sentcount))
print('JA:', sorted(jacor_sentcount))
```

```python
# sentence length, but per-essay measurements. Let's round 'em...
bucor_sentlen = [round(len(bucor.words(f))/len(bucor.sents(f)), 2) for f in bucor.fileids()]
jacor_sentlen = [round(len(jacor.words(f))/len(jacor.sents(f)), 2) for f in jacor.fileids()]
print('BU:', sorted(bucor_sentlen))
print('JA:', sorted(jacor_sentlen))
```

```python
import statistics
print('BU:', statistics.mean(bucor_sentlen), statistics.median(bucor_sentlen))
print('JA:', statistics.mean(jacor_sentlen), statistics.median(jacor_sentlen))
```

### Indicators of writing quality. TTR: Type-Token Ratio
- Type-token ratio (TTR) is a quick measurement for vocabulary diversity.
- Per-group measurement and also per-essay measurement.
- But there's a big caveat. What? 

```python
# per-group measurements
print('BU:', len(bucor_fd)/len(bucor.words()))
print('JA:', len(jacor_fd)/len(jacor.words()))
```

```python
def get_ttr(wlist):
    wlist_lower = [w.lower() for w in wlist]
    return round(len(set(wlist_lower))/len(wlist_lower), 4)

get_ttr('Rose is a rose is a rose is a rose'.split())
```

```python
bucor_ttr = [get_ttr(bucor.words(f)) for f in bucor.fileids()]
jacor_ttr = [get_ttr(jacor.words(f)) for f in jacor.fileids()]
print('BU:', sorted(bucor_ttr))
print('JA:', sorted(jacor_ttr))
```

- OK, those are the numbers... but _how to interpret them_? 
- Also, the big CAVEAT? What do we do about it?

### Indicators of writing quality. Average word length

```python
def get_avg_wlen(wlist):
    wlen = [len(w) for w in wlist]
    return round(sum(wlen)/len(wlen), 2)

get_avg_wlen('Rose is a rose is a rose is a rose'.split())
```

```python
# per-group measurements
print('BU:', get_avg_wlen(bucor.words()))
print('JA:', get_avg_wlen(jacor.words()))
```

```python
bucor_wlen = [get_avg_wlen(bucor.words(f)) for f in bucor.fileids()]
jacor_wlen = [get_avg_wlen(jacor.words(f)) for f in jacor.fileids()]
print('BU:', sorted(bucor_wlen))
print('JA:', sorted(jacor_wlen))
```
